Regression Analysis: Effect of Rewards Membership to Average Customer Spending

```python
import pandas as pd
import statsmodels.api as sm
```

```python
df = pd.read_excel("Regression Data.xlsx")
```

```python
df.info()
```

```python
df.dropna()
```

```python
x = df['Rewards_Member']   # independent variable (0 or 1)
y = df['Average Spending'] # dependent variable
```

```python
x = sm.add_constant(x)
```

```python
model = sm.OLS(y, x).fit()
```

```python
print(model.summary())
```
